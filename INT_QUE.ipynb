### 1. Write a function to reverse a string in Python

```python
def reverse_string(s):
    # Using Python's slice syntax [start:stop:step] with a step of -1 to reverse
    return s[::-1]

# Test the function
print(reverse_string("hello"))
```

### 2. Check if a string is a palindrome.

```python
def is_palindrome(s):
    # A string is a palindrome if it reads the same forward and backward
    return s == s[::-1]

# Test the function
print(is_palindrome("racecar"))
```

### 3. Write a function to find the nth Fibonacci number using recursion.

```python
def fibonacci(n):
    # Base cases: return n if it is 0 or 1
    if n <= 1:
        return n
    # Recursive step
    return fibonacci(n - 1) + fibonacci(n - 2)

# Test the function
n = 5
print(f"The {n}th Fibonacci number is: {fibonacci(n)}")
```

### 4. Find indices of two numbers that add up to a specific target in an array.

```python
def two_nums(nums, target):
    # Store value mapping to its index for O(1) lookups
    num_map = {}
    for index, num in enumerate(nums):
        complement = target - num
        if complement in num_map:
            return [num_map[complement], index]
        num_map[num] = index
    return []

# Test the function
nums = [2, 7, 11, 15]
target = 9
print(two_nums(nums, target))
```

### 5. Write a Python function to calculate the factorial of a number.

```python
def factorial(n):
    # Base cases: 0! and 1! are both 1
    if n == 0 or n == 1:
        return 1
    # Recursive call
    return n * factorial(n - 1)

# Test the function
print(factorial(5))
```

### 6. How would you count the occurrences of each element in a list?

```python
from collections import Counter

def count_occurence(lst):
    # Using Counter class to construct frequency mapping
    return Counter(lst)

# Test the function
print(count_occurence([1, 2, 3, 4, 5, 6, 6, 6, 7, 7]))
```

### 7. Find the first non-repeated character in a string.

```python
def first_non_repeated_char(s):
    char_count = {}
    # First pass: count frequency of each character
    for char in s:
        char_count[char] = char_count.get(char, 0) + 1

    # Second pass: find the first character with a count of 1
    for char in s:
        if char_count[char] == 1:
            return char
    return None

# Test the function
print(first_non_repeated_char("swiss"))
```

### 59. Simulate a biased coin flip using a fair coin function (von Neumann's trick).

```python
import random

def biased_coin():
    # von Neumann's trick: flip a fair coin twice.
    # HT and TH have identical probabilities regardless of bias.
    flip1 = random.randint(0, 1)
    flip2 = random.randint(0, 1)

    # If the outcomes differ, return the first outcome
    if flip1 != flip2:
        return flip1
    # If the outcomes are the same (HH or TT), discard and retry
    return biased_coin()

# Test the function
print(biased_coin())
```

### 58. Write a Python function to calculate the mean, median, and standard deviation of a list of numbers.

```python
import numpy as np

lst = [1, 2, 3, 4, 5, 6, 7]

# Utilizing numpy functions for reliable statistics calculation
mean = np.mean(lst)
median = np.median(lst)
std = np.std(lst)

print(f"The mean is: {mean}")
print(f"The median is: {median}")
print(f"The standard deviation is: {std:.4f}")
```
